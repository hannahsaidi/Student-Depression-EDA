# Part 2: Missing Values and Data Quality Issues

```python
import pandas as pd
import numpy as np

df = pd.read_excel(
    "../../data/Student_Depression_Dataset.xlsx"
)
```

```python
# Count missing values
missing = df.isnull().sum()

# Calculate missing percentage
missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing,
    "Missing Percentage (%)": missing_percentage
})

display(missing_summary)
```

### Missing Values Analysis

The dataset was examined using isnull().sum() to identify missing values in each column.

The inspection showed that all 13 columns contain zero missing values.

Therefore, no rows or columns were removed, and no imputation techniques were required.


```python
# Count duplicate records
print("Total duplicate rows:", df.duplicated().sum())

# Display duplicate rows
display(df[df.duplicated()])
```

### Duplicate Records

The dataset was checked for exact duplicate rows using duplicated(). The inspection found zero duplicate records.

Therefore, no rows needed to be removed due to duplication.

### Data Types

The dataset contains both numerical and categorical attributes.

Numerical attributes include Age, Current_semester, CGPA and Cost_without_study.
